# Configurar o ambiente do projeto para o Projeto Formula1

- Criar Catalog formula1_incr
- Criar Schemas landing, bronze, silver e gold
- Criar Volume Files no schema landing

##Criar o Catalog

```sql
%sql
CREATE CATALOG IF NOT EXISTS formula1_incr;
```

```sql
%sql
SHOW CATALOGS;
```

##Criar os Schemas

```sql
%sql
CREATE SCHEMA IF NOT EXISTS formula1_incr.landing;
CREATE SCHEMA IF NOT EXISTS formula1_incr.bronze;
CREATE SCHEMA IF NOT EXISTS formula1_incr.silver;
CREATE SCHEMA IF NOT EXISTS formula1_incr.gold;
```

```sql
%sql
USE CATALOG formula1_incr;
```

```sql
%sql
SHOW SCHEMAS;
```

##Criar o Volume

```sql
%sql
CREATE VOLUME IF NOT EXISTS formula1_incr.landing.arquivos;
```

```sql
%fs ls /Volumes/formula1_incr/landing/arquivos
```
