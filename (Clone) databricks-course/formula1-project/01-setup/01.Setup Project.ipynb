##Configurando o projeto formula 1

```sql
-- Cria o catalog do projeto
CREATE CATALOG IF NOT EXISTS formula1;

-- Cria os schemas (as camadas do projeto: landing, bronze, silver, gold)
CREATE SCHEMA IF NOT EXISTS formula1.landing;
CREATE SCHEMA IF NOT EXISTS formula1.bronze;
CREATE SCHEMA IF NOT EXISTS formula1.silver;
CREATE SCHEMA IF NOT EXISTS formula1.gold;

-- Cria o Volume onde vão ficar os arquivos brutos (substitui o container do Azure)
CREATE VOLUME IF NOT EXISTS formula1.landing.arquivos;
```

```sql
%python
display(dbutils.fs.ls("/Volumes/formula1/landing/arquivos/"))
```

```sql
%python
df_drivers = spark.read.json("/Volumes/formula1/landing/arquivos/drivers.json")
display(df_drivers)
```
