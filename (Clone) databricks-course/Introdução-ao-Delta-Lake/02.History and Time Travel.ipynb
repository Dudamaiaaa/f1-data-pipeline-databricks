# Histórico e Time Travel

- Consultar o histórico da tabela Delta Lake
- Consultar versões anteriores dos dados
- Consultar dados de um momento específico
- Restaurar dados para uma versão específica

##- Consultar o histórico da tabela Delta Lake

```sql
DESCRIBE HISTORY demo.delta_lake.companies;
```

##- Consultar versões anteriores dos dados

```sql
SELECT * FROM demo.delta_lake.companies
VERSION AS OF 1;
```

##- Consultar dados de um momento específico

```sql
SELECT * FROM demo.delta_lake.companies
TIMESTAMP AS OF '2026-08-17T13:44:54.000+00:00';
```

##- Restaurar dados para uma versão específica

```sql
RESTORE TABLE demo.delta_lake.companies TO VERSION AS OF 1;
```

```sql
SELECT * FROM demo.delta_lake.companies;
```
